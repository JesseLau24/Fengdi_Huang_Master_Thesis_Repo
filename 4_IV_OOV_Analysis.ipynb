```python
import pandas as pd

# load CSV
file_path = r"Summary_CSVs\oov_iv_summary.csv"
df = pd.read_csv(file_path)

# extract pct and model name
df['data_pct'] = df['model'].str.extract(r'_(\d+)pct').astype(int)

# sort the pct
data_order = [1, 5, 20, 50, 100]
tasks = ['POS', 'MORPH', 'UAS', 'LAS']
models = df['model'].str.replace(r'_\d+pct', '', regex=True).unique()

# table
tables = {}

for task in tasks:
    for model_base in models:
        # model and task
        subdf = df[(df['model'].str.startswith(model_base)) & (df['component'] == task)]
        # sort by pct
        subdf = subdf.set_index('data_pct').loc[data_order]
        
        iv = subdf['IV_acc'].values
        oov = subdf['OOV_acc'].values
        gap = iv - oov
        
        table_df = pd.DataFrame([iv, oov, gap],
                                index=['IV', 'OOV', 'Gap'],
                                columns=[f"{p}%" for p in data_order])
        
        tables[f"{model_base}_{task}"] = table_df

# display table
for name, table in tables.items():
    print(f"\n=== {name} ===")
    print(table)
```
